```python
#problem1
undirected = {
    "A": {"B", "C"},
    "B": {"A"},
    "C": {"A"},
    "D": set()
}

vertices = undirected.keys()
inverse = {v: set() for v in vertices}

for u in vertices:
    for v in vertices:
        if u != v:
            if v not in undirected[u]:
                inverse[u].add(v)

print("Original undirected:", undirected)
print("Inverse:", inverse)
```

```python
directed1 = {
    "A": ["B"],
    "B": ["C"],
    "C": []
}

transpose1 = {
    "A": [],
    "B": ["A"],
    "C": ["B"]
}

directed2 = {
    "X": ["Y", "Z"],
    "Y": ["Z"],
    "Z": []
}

transpose2 = {
    "X": [],
    "Y": ["X"],
    "Z": ["X", "Y"]
}

print("Directed graph 1:", directed1)
print("Transpose 1:", transpose1)
print("Directed graph 2:", directed2)
print("Transpose 2:", transpose2)
```

```python
#Problem2
def bron_kerbosch(R, P, X, graph, cliques):
    if not P and not X:
        cliques.append(R.copy())
        return

    for v in list(P):
        bron_kerbosch(
            R | {v},
            P.intersection(graph[v]),
            X.intersection(graph[v]),
            graph,
            cliques
        )
        P.remove(v)
        X.add(v)


graph = {
    "A": {"B", "C"},
    "B": {"A", "C"},
    "C": {"A", "B", "D"},
    "D": {"C"}
}

cliques = []
P = set(graph.keys())
bron_kerbosch(set(), P, set(), graph, cliques)

print("Maximal cliques:", cliques)
```
