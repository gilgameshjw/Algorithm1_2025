```python
import numpy as np
import networkx as nx

INDEX_TO_NAME = {0: 'A', 1: 'B', 2: 'C', 3: 'D', 4: 'E'}
NUM_VERTICES = len(INDEX_TO_NAME)

G1_col_pointers = [0, 2, 5, 8, 11, 12]
G1_row_indices = [1, 2, 0, 2, 3, 0, 1, 3, 1, 2, 4, 3]

Adj_G1 = np.zeros((NUM_VERTICES, NUM_VERTICES), dtype=int)
for j in range(NUM_VERTICES):
    start = G1_col_pointers[j]
    end = G1_col_pointers[j+1]
    for k in range(start, end):
        i = G1_row_indices[k]
        Adj_G1[i, j] = 1

G2_col_pointers = [0, 0, 2, 4, 5, 7]
G2_row_indices = [0, 3, 0, 1, 2, 1, 3]

Adj_G2 = np.zeros((NUM_VERTICES, NUM_VERTICES), dtype=int)
for j in range(NUM_VERTICES):
    start = G2_col_pointers[j]
    end = G2_col_pointers[j+1]
    for k in range(start, end):
        i = G2_row_indices[k]
        Adj_G2[i, j] = 1

print("=== Results for Graph 1 (Undirected) ===")

print("\n(a) Adjacency Matrix for Graph 1 (Adj_G1):")
print("   A B C D E")
for i in range(NUM_VERTICES):
    row_label = INDEX_TO_NAME[i]
    row_str = " ".join(map(str, Adj_G1[i]))
    print(f"{row_label}: {row_str}")

edges_G1 = set()
for j in range(NUM_VERTICES):
    start = G1_col_pointers[j]
    end = G1_col_pointers[j+1]
    for k in range(start, end):
        i = G1_row_indices[k]
        u, v = sorted((INDEX_TO_NAME[i], INDEX_TO_NAME[j]))
        edges_G1.add(tuple(v))

print("\n(b) Edges of Graph 1 for diagram construction:")
print(edges_G1)
print("")

print("\n" + "="*50 + "\n")

print("=== Results for Graph 2 (Directed) ===")

print("\n(a) Adjacency Matrix for Graph 2 (Adj_G2):")
print("   A B C D E")
for i in range(NUM_VERTICES):
    row_label = INDEX_TO_NAME[i]
    row_str = " ".join(map(str, Adj_G2[i]))
    print(f"{row_label}: {row_str}")

edges_G2 = []
for j in range(NUM_VERTICES):
    start = G2_col_pointers[j]
    end = G2_col_pointers[j+1]
    for k in range(start, end):
        i = G2_row_indices[k]
        edges_G2.append((INDEX_TO_NAME[j], INDEX_TO_NAME[i]))

print("\n(b) Edges of Graph 2 for diagram construction (Source -> Target):")
print(edges_G2)
print("")

G2_nx = nx.DiGraph()
G2_nx.add_edges_from(edges_G2)
cycles = list(nx.simple_cycles(G2_nx))

unique_cycle_g2 = ""
if cycles:
    for cycle in cycles:
        if set(cycle) == {'B', 'C', 'E'} and len(cycle) == 3:
            start_index = cycle.index('B')
            ordered_cycle = cycle[start_index:] + cycle[:start_index]
            ordered_cycle_str = "->".join(ordered_cycle) + "->" + ordered_cycle[0]
            if ordered_cycle_str == "B->E->C->B":
                 unique_cycle_g2 = ordered_cycle_str
                 break

print("\n(c) The unique cycle in Directed Graph 2:")
print(unique_cycle_g2 if unique_cycle_g2 else "Cycle not found or different from B->E->C->B")
```

1. Tree Definitions (Problem 1)

The task is to prove that seven different definitions of a tree are mathematically equivalent. The core concept is that a tree is a graph that is connected and acyclic (contains no cycles).



Key equivalent properties include having exactly V−1 edges (for V vertices) , being minimally connected (removing any edge disconnects it) , and being maximally acyclic (adding any edge creates a cycle). A tree also guarantees a unique path between every pair of vertices.
