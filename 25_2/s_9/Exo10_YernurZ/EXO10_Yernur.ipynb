```python
import math

print("--- PROBLEM 1: CLASSIFICATION ---")
print("1. Find max, linear search, shortest path in unweighted graph, matrix multiplication:")
print("      → Class P (Polynomial time)")
print("2. Sorting, Dijkstra (non-negative weights), BFS, DFS, merge sort, quicksort:")
print("      → Class P (Polynomial time)")
print("3. Sudoku:")
print("      → NP-Complete")
print("4. 3-coloring, Scheduling with conflicts:")
print("      → NP-Complete")
print("5. TSP, Hamiltonian Cycle, Clique:")
print("      → NP-Hard (decision versions are NP-Complete)")
print("6. Cryptography, Factoring:")
print("      → In NP (not known to be in P, not proven NP-Complete)")
print("7. Halting Problem, Busy Beaver:")
print("      → Undecidable (cannot be computed)")
print("\n" + "="*30 + "\n")


print("--- PROBLEM 2: BAYES THEOREM ---")

p_disease = 0.001
p_healthy = 1 - p_disease

sensitivity = 0.99
specificity = 0.99

p_false_positive = 1 - specificity

p_true_pos = p_disease * sensitivity
p_false_pos = p_healthy * p_false_positive
total_pos = p_true_pos + p_false_pos

result = p_true_pos / total_pos

print(f"Probability the patient actually has the disease: {result:.4f}")
print(f"As a percentage: {result * 100:.2f}%")
print("\nWhy is this only ~9%?")
print("Because the disease is extremely rare (0.1%).")
print("Even though the test is accurate, false positives from healthy people")
print("are much more common than true positives from sick people.")
print("\n" + "="*30 + "\n")


print("--- PROBLEM 3: SHANNON ENTROPY ---")

def entropy(p):
    if p == 0 or p == 1:
        return 0
    q = 1 - p
    return -(p * math.log2(p) + q * math.log2(q))

coins = [
    ("Coin A (50%)", 0.50),
    ("Coin B (99%)", 0.99),
    ("Coin C (1%)", 0.01)
]

for name, p in coins:
    h = entropy(p)
    print(f"{name}: P(Heads)={p}, Entropy={h:.4f} bits")

print("\nWhy is the fair coin 1 bit?")
print("Because outcomes are fully unpredictable (50/50). It requires 1 full bit.")
print("\nWhy is the biased coin ~0.08 bits?")
print("Because the result is almost always known in advance (e.g., 99% Heads).")
print("Less uncertainty → fewer bits of information.")
```
